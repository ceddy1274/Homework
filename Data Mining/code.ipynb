# Load data

```python
import pandas as pd
import numpy as np
from sklearn import datasets
```

## Iris dataset

```python
iris = datasets.load_iris(as_frame=True)
```

```python
iris['data']
```

```python
iris['target']
```

```python
iris_labels = iris['target']
iris_label_names = iris['target_names']
iris = iris['data']

display(iris)
display(iris_labels)
print(iris_label_names)
```

```python
iris.describe()
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split

train_iris, test_iris, train_iris_labels, test_iris_labels = train_test_split(
    iris,
    iris_labels,
    test_size=0.30,
    random_state=42,
    shuffle=True)

print(len(train_iris), len(test_iris))
```

## Breast Cancer dataset

```python
cancer = datasets.load_breast_cancer(as_frame=True)
cancer_labels = cancer['target']
cancer_label_names = cancer['target_names']
cancer = cancer['data']

display(cancer)
display(cancer_labels)
print(cancer_label_names)
```

```python
cancer.describe()
```

### Train Test Split

```python
train_cancer, test_cancer, train_cancer_labels, test_cancer_labels = train_test_split(
    cancer,
    cancer_labels,
    test_size=0.30,
    random_state=42,
    shuffle=True)
```

```python
print(len(train_cancer), len(test_cancer))
```

### Data Normalization

```python
from sklearn import preprocessing

scaler = preprocessing.StandardScaler()
scaler.fit(train_cancer)
train_cancer[train_cancer.columns] = scaler.transform(train_cancer)

train_cancer.describe()
```

```python
test_cancer[test_cancer.columns] = scaler.transform(test_cancer)
```

```python
train_cancer, test_cancer, train_cancer_labels, test_cancer_labels = train_test_split(
    cancer,
    cancer_labels,
    test_size=0.30,
    random_state=42,
    shuffle=True)

minmaxscaler = preprocessing.MinMaxScaler(feature_range=(0,1))
minmaxscaler.fit(train_cancer)
train_cancer[train_cancer.columns] = minmaxscaler.transform(train_cancer)
test_cancer[test_cancer.columns] = minmaxscaler.transform(test_cancer)

train_cancer.describe()
```

# Classification in Iris

## Decision Trees

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

dtree = DecisionTreeClassifier()
dtree.fit(train_iris, train_iris_labels)
predictions = dtree.predict(test_iris)
print(dtree.get_n_leaves())

r = tree.export_text(dtree)
print(r)

tree.plot_tree(dtree)
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report

print("Accuracy: ", accuracy_score(test_iris_labels, predictions)*100)
print(classification_report(test_iris_labels, predictions, target_names=iris_label_names))
```

```python
# Visualize confusion matrix
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(test_iris_labels, predictions, labels=dtree.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=iris_label_names)
disp.plot()
```

# Classification on Cancer

```python
dtree = DecisionTreeClassifier()
dtree.fit(train_cancer, train_cancer_labels)
predictions = dtree.predict(test_cancer)
print(dtree.get_n_leaves())

r = tree.export_text(dtree)
print(r)

tree.plot_tree(dtree)
plt.show()
```

```python
print("Accuracy: ", accuracy_score(test_cancer_labels, predictions)*100)
print(classification_report(test_cancer_labels, predictions, target_names=cancer_label_names))
```

```python
cm = confusion_matrix(test_cancer_labels, predictions, labels=dtree.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=cancer_label_names)
disp.plot()
```

## Visualization

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca.fit(train_cancer)
test_pca_cancer = pca.transform(test_cancer)
```

```python
# Plot predictions
plt.figure()
fig,ax =plt.subplots(1,2,figsize=(12,5), sharex=True, sharey=True)

length = ax[0].scatter(x=test_pca_cancer[:,0], y=test_pca_cancer[:,1], c=test_cancer_labels)
width = ax[1].scatter(x=test_pca_cancer[:,0], y=test_pca_cancer[:,1], c=predictions)

ax[0].set_title('True')
ax[1].set_title('Predicted')
ax[0].set_xlabel('PCA 1')
ax[1].set_xlabel('PCA 1')
ax[0].set_ylabel('PCA 2')
handles, labels = length.legend_elements()
legend = ax[0].legend(handles = handles, labels = list(cancer_label_names), title="Classes")
legend = ax[1].legend(handles = handles, labels = list(cancer_label_names), title="Classes")

plt.show()
```

# Clustering on Cancer

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=2, random_state=0, n_init="auto")
kmeans.fit(train_cancer)

print(kmeans.cluster_centers_)
print(kmeans.labels_)
```

```python
predicted_cluster_labels = kmeans.predict(test_cancer)
predicted_cluster_labels
```

```python
plt.figure()
fig,ax =plt.subplots(1,2,figsize=(12,5), sharex=True, sharey=True)

length = ax[0].scatter(x=test_pca_cancer[:,0], y=test_pca_cancer[:,1], c=test_cancer_labels)
width = ax[1].scatter(x=test_pca_cancer[:,0], y=test_pca_cancer[:,1], c=predicted_cluster_labels)

ax[0].set_title('True')
ax[1].set_title('Predicted')
ax[0].set_xlabel('PCA 1')
ax[1].set_xlabel('PCA 1')
ax[0].set_ylabel('PCA 2')
handles, labels = length.legend_elements()
legend = ax[0].legend(handles = handles, labels = list(cancer_label_names), title="Classes")
legend = ax[1].legend(handles = handles, labels = list(cancer_label_names), title="Classes")

plt.show()
```

```python
cm = confusion_matrix(test_cancer_labels, predicted_cluster_labels, labels=dtree.classes_)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=cancer_label_names)
disp.plot()
```
